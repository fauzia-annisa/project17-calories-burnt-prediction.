Importing the Dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn import metrics
```

Data Collection & Processing

```python
# loading the data from csv file to a Pandas DataFrame
calories = pd.read_csv('/content/calories.csv')
```

```python
# print the first 5 rows of the dataframe
calories.head()
```

```python
exercise_data = pd.read_csv('/content/exercise.csv')
```

```python
exercise_data.head()
```

Combining the two Dataframes

```python
calories_data = pd.concat([exercise_data, calories['Calories']], axis=1)
```

```python
calories_data.head()
```

```python
# checking the number of rows and columns
calories_data.shape
```

```python
# getting some informations about the data
calories_data.info()
```

```python
# checking for missing values
calories_data.isnull().sum()
```

Data Analysis

```python
# get some statistical measures about the data
calories_data.describe()
```

Data Visualization

```python
sns.set()
```

```python
# plotting the gender value in vount plot
sns.set_theme(style="whitegrid")
BABY_PINK = '#FFB6C1'
LIGHT_PINK = '#FFE4E1'

plt.figure(figsize=(8,5))
sns.countplot(y=calories_data['Gender'], color=BABY_PINK)
plt.xlabel('count', color='#555')
plt.ylabel('Gender', color='#555')
plt.title('Gender Distribution', color='#555')
plt.show()
```

```python
# finding the distribution of "Age" column
plt.figure(figsize=(8,5))
sns.histplot(calories_data['Age'], kde=True, color=BABY_PINK, line_kws={'color':'#FF69B4', 'lw':2})
plt.xlabel('Age', color='#555')
plt.ylabel('Density', color='#555')
plt.title('Age Distribution', color='#555')
plt.show()
```

```python
# finding the distribution of "Height" column
sns.histplot(calories_data['Height'], kde=True, stat='density', color='#F8C8DC')
```

```python
# finding the distribution of "Weight" column
sns.histplot(calories_data['Weight'], kde=True, stat='density', color='#F8C8DC')
```

Finding the Correlation in the dataset

1. Positive Correlation
2. Negative Correlation

```python
correlation = calories_data.corr()
```

```python
from matplotlib.colors import LinearSegmentedColormap

# custom colormap
baby_pink = LinearSegmentedColormap.from_list(
    'baby_pink', ['#FFF5F8', '#F8C8DC', '#E58AAE'])

# constructing a heatmap to understand the correlation
plt.figure(figsize=(10,10))
sns.heatmap(correlation, cbar=True, square=True, fmt='.1f', annot=True,
                annot_kws={'size':8}, cmap=baby_pink)
```

Converting the text data to numerical values

```python
calories_data.replace({"Gender":{'male':0,'female':1}}, inplace=True)
```

```python
calories_data.head()
```

Separating features and Target

```python
X = calories_data.drop(columns=['User_ID','Calories'], axis=1)
Y = calories_data['Calories']
```

```python
print(X)
```

```python
print(Y)
```

Splitting the data into training data and Test data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Model Training

XGBoost Regressor

```python
# loading the model
model = XGBRegressor()
```

```python
# training the model with X_train
model.fit(X_train, Y_train)
```

Evaluation

Prediction on Test Data

```python
test_data_prediction = model.predict(X_test)
```

```python
print(test_data_prediction)
```

Mean Absolute Error

```python
mae = metrics.mean_absolute_error(Y_test, test_data_prediction)
```

```python
print("Mean Absolute Error = ", mae)
```
